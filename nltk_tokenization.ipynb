```python
paragraph = "Natural Language Processing (NLP) is a field of artificial intelligence that helps computers understand human language. Today, NLP powers translation apps, voice assistants, spam filters, and chatbots. Dr. Smith's team in New York isn't finished: they are studying how models read, running thousands of experiments every day! Their results were better than expected, and the cities of Casablanca and Rabat showed great interest. Can machines truly understand us? Visit https://example.com or email me at test@example.com #NLP :)" 
print("ORIGINAL PARAGRAPH:\n", paragraph)
print("\nTotal characters:", len(paragraph))
```

# ------------------------------------------------------------
# 2) SENTENCE TOKENIZATION
# ------------------------------------------------------------

```python
import nltk
# nltk.download("punkt_tab")
for resource in ["punkt", "punkt_tab", "stopwords", "wordnet",
                 "averaged_perceptron_tagger", "averaged_perceptron_tagger_eng"]:
    nltk.download(resource, quiet=True) 
```

```python
from nltk.tokenize import sent_tokenize
```

```python
# Split the paragraph into a list of sentences
sentence = sent_tokenize(paragraph)
```

```python
print("\n--- SENTENCE TOKENIZATION ---")
print("Number of sentences:", len(sentence))
```

```python
for i, s in enumerate(sentence, start=1):
    print(f"{i} {s}")
```

```python
# ------------------------------------------------------------
# 3) WORD TOKENIZATION (the standard way)
# ------------------------------------------------------------
```

```python
from nltk.tokenize import word_tokenize
```

```python
# Split the whole paragraph into word tokens
tokens = word_tokenize(paragraph)
```

```python
print(f"Number of tokens : {len(tokens)}")
print(tokens)
```

# ------------------------------------------------------------
# 4) OTHER TOKENIZERS: SAME TEXT, DIFFERENT RESULTS
# ------------------------------------------------------------

```python
from nltk.tokenize import WordPunctTokenizer, TweetTokenizer, RegexpTokenizer
```

```python
# A short text with tricky parts
sample = "Dr. Smith isn't here! Visit https://example.com #NLP :)"
print("\n--- COMPARING TOKENIZERS ON:", sample)
```

```python
# Standard: splits "isn't" into "is" and "n't"
print(f"word_tokenize : {word_tokenize(sample)}")
print("="*141)
# Splits every punctuation symbol apart
print(f"WordPunctTokenizer : {WordPunctTokenizer().tokenize(sample)}")
print("="*141)
# Keeps URLs, hashtags and emoticons whole
print(f"TweetTokenizer : {TweetTokenizer().tokenize(sample)}")
print("="*141)
# Plain Python split on spaces (the naive way)
print(f"WhiteSpace : {sample.split()}")
```

```python
# Custom rule: keep only runs of letters/digits/underscore
word_only = RegexpTokenizer(r"\w+")
print("Regexp (\\w+)  :", word_only.tokenize(sample))
```

# ------------------------------------------------------------
# 5) CLEANING: LOWERCASE, REMOVE PUNCTUATION AND STOP WORDS
# ------------------------------------------------------------

```python
from nltk.corpus import stopwords
```

```python
# Make every token lowercase
lowered = [t.lower() for t in tokens]
```

```python
# Keep tokens made only of letters (drops punctuation, numbers)
only_words = [t for t in lowered if t.isalpha()]
```

```python
# Load English stop words into a set (fast lookup)
stop_set = set(stopwords.words("english"))
```

```python
# Remove the stop words
no_stop = [t for t in only_words if t not in stop_set]
```

```python
print("\n--- CLEANING ---")
# How many tokens remain after removing punctuation
print("Letters only:", len(only_words), "tokens")

# How many remain after removing stop words
print("After removing stop words:", len(no_stop), "tokens")
print("-"*20)
print(no_stop)
```
